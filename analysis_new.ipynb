```python
import numpy as np
import pandas as pd

df_users = pd.read_csv("users.csv", parse_dates=["reg_date"])
df_transactions = pd.read_csv("transactions.csv", parse_dates=["tx_date"])
df_kyc = pd.read_csv("kyc.csv", parse_dates=["kyc_date"])

print("Данные успешно загружены")
print(f"Всего пользователей: {len(df_users)}\nВсего транзакций: {len(df_transactions)}")
```

```python
# удаляем полные дубликаты
df_clean = df_transactions.drop_duplicates()

# удаляем транзакции без МСС-кода
df_clean = df_clean.dropna(subset=["mcc_code"])

# убираем транзакции из прошлого
df_clean = df_clean.merge(df_users[["user_id", "reg_date"]], on="user_id")
df_clean = df_clean[df_clean["tx_date"] >= df_clean["reg_date"]]

print(f"Очистка завершена!")
print(f"Было строк: {len(df_transactions)}")
print(f"Стало строк: {len(df_clean)}")
print(f"Удалено: {len(df_transactions) - len(df_clean)}")
```

```python
# RFM-анализ (сегментация клиентов)


def rfm_analysis(df):
    """
    Recency: как давно была последняя транзакция (чем меньше дней, тем лучше)
    Frequency: сколько всего транзакций (чем больше, тем лучше)
    Monetary: сколько всего потратил (чем больше, тем лучше)
    """
    last_date = df["tx_date"].max()


    rfm = df.groupby("user_id").agg({
        "tx_date" : lambda x: (last_date - x.max()).days,
        "tx_id" : "count",
        "amount" : "sum"
    }).rename(columns={
        "tx_date" : "recency",
        "tx_id" : "frequency",
        "amount" : "monetary"
    })


    rfm["R"] = pd.qcut(rfm["recency"], 4, labels=["4", "3", "2", "1"])  # 4 = недавно
    rfm["F"] = pd.qcut(rfm["frequency"], 4, labels=["1", "2", "3", "4"])  # 4 = часто
    rfm["M"] = pd.qcut(rfm["monetary"], 4, labels=["1", "2", "3", "4"])  # 4 = много


    rfm["RFM_Score"] = rfm["R"].astype(str) + rfm["F"].astype(str) + rfm["M"].astype(str)

    # сегменты по RFM
    segments = {"444": "💎 Топ-актив (недавно, часто, много)",
        "441": "📈 Активные микрочеки (недавно, часто, мало)",
        "414": "🐳 Разовые киты (недавно, редко, много)",
        "344": "⚠️ Уходящая лояльность (затишье, часто, много)",
        "144": "🆘 Потерянные VIP (давно, часто, много)",
        "211": "📉 Уснувшие новички (давно, редко, мало)",
        "411": "🌱 Свежая кровь (недавно, редко, мало)",
        "111": "💤 Мертвая база (очень давно, редко, мало)"
        }


    rfm["Segment"] = rfm["RFM_Score"].map(segments).fillna("Другие")
    return rfm


rfm_results = rfm_analysis(df_clean)

print("--- RFM-АНАЛИЗ КЛИЕНТОВ ---")
print(f"Всего клиентов: {len(rfm_results)}")

# топ-сегменты
top_segments = rfm_results["Segment"].value_counts()

for seg, count in top_segments.items():
    print(f"{seg} : {count} клиентов ({count/len(rfm_results)*100:.1f}%)")

# БИЗНЕС-ВЫВОД
print(f"\n--- БИЗНЕС-ВЫВОД ---")
print("💎 ГЛАВНЫЙ РИСК: Отток VIP-клиентов (4.5%) почти равен их притоку (4.8%).")
print("Инсайт: Мы теряем базу лояльных клиентов так же быстро, как выращиваем новую.")
print("Выручка от Потерянных VIP (7.6%) практически идентична выручке Топов (7.8%).")

# про удержание клиентов
print("\n🆘 УДЕРЖАНИЕ: Работа с Уходящей лояльностью (4.3%) эффективнее привлечения новичков.")
print("Инсайт: Стоимость удержания текущих VIP ниже стоимости маркетинга для Свежей крови.")

# экономическая ценность
print("\n💤 ЭКОНОМИЯ: Скидка 20% для Мертвой базы нецелесообразна из-за их низкого чека.")
print("Инсайт: Скидка для этого сегмента приведет к убыткам и уничтожению маржи.")

# поиск точек роста
print("\n📈 ТОЧКА РОСТА: В сегменте Другие (71% базы) скрыт основной потенциал масштабирования и роста.")
print("Инсайт: Необходим глубокий ABC-анализ внутри группы для поиска будущих драйверов прибыли.")

# индивидуальный подход
print("\n🐳 ПРИОРИТЕТ: 6 Разовых китов требуют личного менеджмента для повторных продаж.")
print("Инсайт: Конверсия всего 1-2 таких клиентов в постоянные даст рост выручки выше, чем сотни мелких чеков.")

champions_revenue = rfm_results[rfm_results["RFM_Score"] == "444"]["monetary"].sum()
total_revenue = rfm_results["monetary"].sum()

print("---\n ТОП-СЕГМЕНТЫ ПО ВЫРУЧКЕ ---")
print(f"Топ-актив (4.8%): {champions_revenue/total_revenue*100:.1f}% выручки")
print(f"Потерянные VIP (4.5%): {rfm_results[rfm_results['RFM_Score'] == '144']['monetary'].sum()/total_revenue*100:.1f}% выручки")

# ВИЗУАЛИЗАЦИЯ
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams['font.family'] = ['Arial', 'Segoe UI Emoji', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False
segment_stats = rfm_results.groupby('Segment').agg(
    clients=('monetary', 'count'),
    revenue=('monetary', 'sum')
).reset_index()

segment_stats["client_share"] = segment_stats["clients"] / len(rfm_results) * 100
segment_stats["revenue_share"] = segment_stats["revenue"] / segment_stats["revenue"].sum() * 100
segment_stats = segment_stats.sort_values('client_share', ascending=False)

x = range(len(segment_stats))
width = 0.35

plt.figure(figsize=(14,6))
plt.bar([i - width/2 for i in x], segment_stats["client_share"], width, label="% клиентов")
plt.bar([i + width/2 for i in x], segment_stats["revenue_share"], width, label="% выручки")
plt.xticks(x, segment_stats['Segment'], rotation=45, ha='right')
plt.legend()
plt.title("Доля клиентов и выручки по сегментам")
plt.tight_layout()

plt.savefig('rfm_analysis.png', dpi=300, bbox_inches='tight')   # RFM
plt.show()
```

```python

# LTV-АНАЛИЗ. СКОЛЬКО ДЕНЕГ ПРИНОСИТ КЛИЕНТ
df_clean['net_revenue'] = df_clean['amount'] - df_clean['cashback_amount']

ltv_user = df_clean.groupby("user_id")["net_revenue"].sum().reset_index()
ltv_user.columns = ["user_id", "ltv"]

print("--- LTV-анализ ---")
print(f"Всего клиентов: {len(ltv_user)}")
print(f"Средний LTV: {ltv_user["ltv"].mean():.1f}")
print(f"Медианный LTV: {ltv_user["ltv"].median():.1f}")
print(f"Минимальный LTV: {ltv_user["ltv"].min():.1f}")
print(f"Максимальный LTV: {ltv_user["ltv"].max():.1f}")

print(f"\n--- КВАРТИЛИ ---")
print(f"25% клиентов приносят меньше: {ltv_user["ltv"].quantile(0.25):.1f}")
print(f"50% клиентов приносят меньше: {ltv_user["ltv"].quantile(0.5):.1f}")
print(f"75% клиентов приносят меньше: {ltv_user["ltv"].quantile(0.75):.1f}")
print(f"ТОП-10% клиентов приносят от: {ltv_user["ltv"].quantile(0.9):.1f}")


ltv_by_segment = ltv_user.merge(
    rfm_results[["Segment"]],
    left_on="user_id",
    right_index=True
)

print(f"\n--- LTV ПО СЕГМЕНТАМ ---")
segment_ltv = ltv_by_segment.groupby("Segment")["ltv"].agg(["mean", "count", "sum"]).round(1)
segment_ltv["доля клиентов"] = (segment_ltv["count"] / len(ltv_user) * 100).round(1)
segment_ltv["доля прибыли"] = (segment_ltv["sum"] / ltv_user["ltv"].sum() * 100).round(1)
print(segment_ltv.sort_values("mean", ascending=False))



# БИЗНЕС-ВЫВОД ПО LTV
print(f"\n--- БИЗНЕС-ВЫВОД ---")
print("Пока мы фокусируемся на других сегментах, 3545 человек из сегмента 'Другие' живут без персонализированного маркетинга и могут уйти к конкурентам")
# про привлечение новых
print("\nСКОЛЬКО МОЖНО ТРАТИТЬ НА ОДНОГО НОВИЧКА:")
print(f"Безопасная сумма на привлечение — 33 036 руб.(30% от медианы {110120.7:.0f})..")
print("Инсайт: Лучше ориентироваться на медиану (110к), а не на среднее (138к),")
print(f"чтобы не переплатить за рекламу из-за случайных крупных заказов.\n")

# про важность старичков
print("ПОЧЕМУ УДЕРЖИВАТЬ СТАРЫХ ВЫГОДНЕЕ:")
print("Один постоянный клиент из Топ-актива (224к) приносит в 3.5 раза больше,")
print("чем один Новичок (63к).")
print("Инсайт: Если мы просто сохраним 216 человек из Уходящих,")
print(f"то спасем 47.5 млн руб. выручки без затрат на новую рекламу.\n")

# про Китов
print("ЧТО ДЕЛАТЬ С 6 САМЫМИ КРУПНЫМИ КЛИЕНТАМИ:")
print("У Китов средний чек — 884к. Это в 6.5 раз выше, чем у обычных людей.")
print("Инсайт: Нам нельзя их терять. Каждому нужно позвонить лично и предложить")
print(f"особые условия, чтобы они покупали снова.\n")

# где мы теряем деньги
print("ГДЕ МОЖНО СЭКОНОМИТЬ БЮДЖЕТ:")
print("Мертвая база и Уснувшие (почти 10% всех клиентов) дают всего 7% прибыли.")
print("Инсайт: Не стоит тратить на них дорогой маркетинг. Лучше направить")
print("эти деньги на тех, кто приносит основные деньги (VIP-клиенты).")

# ВИЗУАЛИЗАЦИЯ

plt.rcParams["font.family"] = ["Arial", "Segoe UI Emoji", "DejaVu Sans"]


stats = ltv_by_segment.groupby("Segment")["ltv"].agg(["count", "sum"])
stats["% клиентов"] = stats["count"] / stats["count"].sum() * 100
stats["% выручки"] = stats["sum"] / stats["sum"].sum() * 100
stats = stats.sort_values("% клиентов", ascending=False)

stats[["% клиентов", "% выручки"]].plot(kind="bar", figsize=(12, 5), width=0.8, color=["#a8dadc", "#e63946"])

plt.title("Доля клиентов и выручки по сегментам")
plt.xticks(rotation=45, ha="right")
plt.ylabel("Процент (%)")
plt.tight_layout()

plt.savefig('ltv_analysis.png', dpi=300, bbox_inches='tight')   # LTV
plt.show()
```

```python
# АНАЛИЗ UNIT-ЭКОНОМИКИ
if 'segment' not in df_clean.columns:
    df_clean = df_clean.merge(df_users[['user_id', 'segment']], on='user_id', how='left')


df_clean["interchange_revenue"] = df_clean["amount"] * 0.015

# считаем чистую прибыль по каждой транзакции
df_clean["net_revenue"] = df_clean["interchange_revenue"] - df_clean["cashback_amount"]


economics_by_segment = df_clean.groupby("segment").agg({
    "user_id" : "nunique",
    "amount" : "sum",
    "cashback_amount" : "sum",
    "net_revenue" : "sum"
}).rename(columns={"user_id" : "users_count", "amount" : "total_volume"})

# считаем ARPU - средняя прибыль на одного пользователя в каждом сегменте
economics_by_segment["ARPU"] = economics_by_segment["net_revenue"] / economics_by_segment["users_count"]

print("--- ЭКОНОМИКА ПО СЕГМЕНТАМ ---")
print(economics_by_segment[["users_count", "net_revenue", "ARPU"]].round(2))

# БИЗНЕС-ВЫВОД
print(f"\n--- БИЗНЕС-ВЫВОД ---")
# проблема Premium-сегмента
print("1. ПАРАДОКС PREMIUM-ПАКЕТА:")
print("Инсайт: ARPU сегмента Premium (148.4) в 6.5 раз НИЖЕ, чем у Standard (963.9).")
print("Вывод: Название Premium не соответствует финансовой реальности.")
print("Решение: Срочно пересмотреть состав пакета. Сейчас клиенты Standard приносят")
print(f"больше денег, имея, вероятно меньше привилегий.\n")

# НАША ОСНОВНАЯ СИЛА
print("ГЛАВНЫЙ ГЕНЕРАТОР ПРИБЫЛИ - Standard:")
print("Инсайт: Сегмент приносит 3.45 млн руб. чистой прибыли.")
print("Вывод: Это наше ядро. Любое изменение цен здесь может сильно ударить по бизнесу.")
print(f"Решение: Использовать Standard как базу для предложения к переходу в VIP-сегмент.\n")

# эффективность VIP
print("VIP-ЭФФЕКТИВНОСТЬ:")
print(" Инсайт: У VIP самый высокий доход на 1 клиента (1159.7 ARPU).")
print("Вывод: Сегмент приносит в 8 раз больше прибыли на человека, чем Premium.")
print("Решение: Программа 'Standard -> VIP' должна стать приоритетом для роста прибыли.")

# ВИЗУАЛИЗАЦИЯ

plt.figure(figsize=(10, 6))

ax = sns.barplot(
    data=economics_by_segment.reset_index(),
    x='segment',
    y='ARPU',
    hue='segment',
    palette='magma',
    legend=False
)

plt.title("Сравнение ARPU по сегментам", fontsize=16, pad=20)
plt.xlabel("Сегмент", fontsize=12)
plt.ylabel("Средняя прибыль на клиента (ARPU)", fontsize=12)

for p in ax.patches:
    ax.annotate(f'{p.get_height():.1f}',
                (p.get_x() + p.get_width() / 2., p.get_height()),
                ha='center', va='baseline', fontsize=11, fontweight='bold', xytext=(0, 5),
                textcoords='offset points')

plt.tight_layout()

plt.savefig('unit_economy_arpu.png', dpi=300, bbox_inches='tight')  # UNIT-ЭКОНОМИКА
plt.show()
```

```python
# АНАЛИЗ УБЫТОЧНОСТИ
query_mcc  = """
SELECT
    mcc_code,
    COUNT(tx_id) as total_tx,
    ROUND(SUM(amount), 2) as total_volume,
    ROUND(SUM(cashback_amount), 2) as total_cashback,
    ROUND(SUM(amount * 0.015) - SUM(cashback_amount), 2) as net_profit
FROM transactions
WHERE status = "Success"
GROUP BY mcc_code
ORDER BY net_profit ASC
LIMIT 10;
"""


mcc_losses = pd.read_sql(query_mcc, conn)
print("--- ТОП-10 УБЫТОЧНЫХ КАТЕГОРИЙ---")
print(mcc_losses)

# БИЗНЕС-ВЫВОД

print(f"\n--- БИЗНЕС-ВЫВОД ---")
# самая убыточная категория
print("УБЫТОЧНЫЕ РЕСТОРАНЫ (MCC 5812):")
print("Инсайт: Единственная категория с чистым убытком (-419 341 руб) при огромном обороте.")
print("Вывод: Мы тратим на кэшбэк больше, чем зарабатываем на комиссиях в этой категории. Выплачиваем в 2 раза больше, чем в других категориях.")
print("Решение: Пересмотреть процент кэшбэка для ресторанов или ввести лимит на выплаты,\n")
print(f"чтобы категория перестала тянуть общую прибыль вниз.\n")

# низкомаржинальные категории
print("НИЗКАЯ МАРЖА В ПОВСЕДНЕВНЫХ ТРАТАХ (MCC 5411, 4111, 4814):")
print("Инсайт: Огромный объем транзакций (более 26 000 суммарно), но очень низкая доходность.")
print("Вывод: Это категории, которые удерживают клиента, но почти не приносят денег.")
print("Решение: Не снижать лояльность здесь, но внедрять партнерские предложения,\n")
print(f"где кэшбэк будет платить магазин-партнер, а не банк.\n")

# наличные
print("ВЫСОКИЙ ОБОРОТ НАЛИЧНЫХ (MCC 6011):")
print("Инсайт: Оборот по снятию наличных превышает 107 млн рублей.")
print("Вывод: Деньги вымываются из системы, банк теряет возможность зарабатывать на транзакциях.")
print("Решение: Стимулировать переход в безналичный расчет через бонусы за оплату по QR или СБП,")
print("чтобы клиенту было выгоднее платить картой, чем снимать наличные деньги.")

# ВИЗУАЛИЗАЦИЯ

data_plot = mcc_losses.head(5).copy()

data_plot['net_profit'] = data_plot['net_profit'].astype(float)
data_plot['mcc_code'] = data_plot['mcc_code'].astype(str)

plt.figure(figsize=(10, 5))

colors = ['#d9534f' if x < 0 else '#808080' for x in data_plot['net_profit']]

ax = sns.barplot(
    data=data_plot,
    x='net_profit',
    y='mcc_code',
    palette=colors,
    hue='mcc_code',
    legend=False
)

plt.axvline(0, color='black', linewidth=1)

for i, v in enumerate(data_plot['net_profit']):
    ax.text(v, i, f' {int(v):,} р.', va='center', fontweight='bold',
            color='red' if v < 0 else 'black')

plt.title("Убыточные категории (Чистая прибыль)", pad=15)
plt.xlabel("Прибыль / Убыток в рублях")
plt.ylabel("MCC-код")

plt.subplots_adjust(left=0.2)
plt.savefig('mcc_profitability_losses.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
# АНАЛИЗ КЛЮЧЕВЫХ КЛИЕНТОВ
df_kyc.to_sql("kyc", conn, index=False, if_exists="replace")

vip_query = """
SELECT
    u.user_id,
    u.segment,
    k.kyc_status,
    COUNT(t.tx_id) as total_tx,
    ROUND(SUM(t.amount * 0.015)  - SUM(t.cashback_amount), 2) as net_profit
FROM users u
JOIN kyc k ON u.user_id = k.user_id
JOIN transactions t ON u.user_id = t.user_id
WHERE t.status = 'Success'
    AND k.kyc_status = 'Approved'
GROUP BY 1, 2, 3
ORDER BY net_profit DESC
LIMIT 10;
"""

vip_result = pd.read_sql_query(vip_query, conn)
print("--- ТОП-10 САМЫХ ПРИБЫЛЬНЫХ КЛИЕНТОВ ---")
print(vip_result)

# БИЗНЕС-ВЫВОД

print(f"\n--- БИЗНЕС-ВЫВОД ---")

# сегментация
print("ДОМИНИРОВАНИЕ СЕГМЕНТА STANDARD:")
print("Инсайт: 8 из 10 самых прибыльных клиентов — это обычные пользователи, а не VIP.")
print("Вывод: Массовый сегмент генерирует основную прибыль за счет частоты операций,")
print("даже если средний чек там ниже, чем у VIP-клиентов.")
print("Решение: Масштабировать инструменты удержания именно для 'Standard' сегмента,")
print("так как они — наш основной источник дохода.\n")


# эффективность транзакций
print("ВЫСОКАЯ ПРИБЫЛЬ ПРИ МАЛОМ ЧИСЛЕ ОПЕРАЦИЙ:")
print("Инсайт: Топ-клиент (ID 1141) принес почти 8000 руб. всего за 11 транзакций.")
print("Вывод: Есть категория клиентов с очень высоким средним чеком или использованием")
print("высокомаржинальных сервисов (например, переводы с большой комиссией или покупка валюты).")
print("Решение: Изучить поведение таких золотых клиентов и предложить похожим пользователям")
print("аналогичные услуги через персональные предложения.")

# ВИЗУАЛИЗАЦИЯ

vip_result['user_id'] = vip_result['user_id'].astype(str)

plt.figure(figsize=(12, 6))

ax = sns.barplot(
    data=vip_result,
    x='net_profit',
    y='user_id',
    hue='segment',
    palette='viridis'
)

plt.title("Топ-10 самых прибыльных клиентов", fontsize=15, pad=20)
plt.xlabel("Чистая прибыль (руб.)")
plt.ylabel("ID Клиента")
plt.legend(title="Сегмент", loc='lower right')

for p in ax.patches:
    width = p.get_width()
    if width > 0:
        ax.annotate(f'{int(width)} р.',
                    (width, p.get_y() + p.get_height() / 2.),
                    va='center', xytext=(5, 0), textcoords='offset points',
                    fontsize=10, fontweight='bold')

plt.tight_layout()
plt.savefig('top_profitable_clients.png', dpi=300, bbox_inches='tight')
plt.show()
```
